```python
#todo list

tasks= []



def add_task():
    task = input("Enter task: ")
    tasks.append(task)
    print("Task added successfully")
def view_task():
    if len(tasks) ==0:
        print("No task in the list")
    else:
        print("\nYour Tasks: ")
        for index, task in enumerate(tasks, start=1):
            print(index, task)
def update_task():
    view_task()

    if len(tasks)==0:
        return
    number= int(input("Enter your task number to update: "))
    if number >= 1 and number <= len(tasks):
        new_task = input("Enter new task: ")
        tasks[number -1] = new_task
        print("Task updated successfully")
    else:
        print("Invalid task number: " )
def delete_task():
    view_task()

    if len(tasks) ==0:
        return
    number = int(input("Enter task number to delete: "))
    if number >= 1 and number <= len(tasks):
        tasks.pop(number-1)
        print("Task deleted successfully")
    else:
        print("Invalid task number")

while True:
    print("\n---To do List---")
    print("1. Add Task")
    print("2. View Task")
    print("3. Update Task")
    print("4. Delete Task")
    print("5. Exit")

    choice= input("Enter your choice: ")

    if choice == "1":
        add_task()
    elif choice == "2":
        view_task()
    elif choice == "3":
        update_task()
    elif choice =="4":
        delete_task()
    elif choice=="5":
        print("To do list stoped")
        break
    else:
        print("Invalid choice")
```
